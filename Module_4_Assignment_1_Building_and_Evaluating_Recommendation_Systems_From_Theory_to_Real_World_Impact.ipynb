```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Sample dataset (user-item ratings)
data = {
    'Movie1': [5, 4, 0, 0, 1],
    'Movie2': [4, 0, 0, 2, 1],
    'Movie3': [1, 1, 0, 5, 0],
    'Movie4': [0, 0, 5, 4, 0],
    'Movie5': [0, 1, 5, 4, 0],
}

df = pd.DataFrame(data, index=['User1', 'User2', 'User3', 'User4', 'User5'])

# Compute similarity between users
similarity = cosine_similarity(df)
similarity_df = pd.DataFrame(similarity, index=df.index, columns=df.index)

# Recommend movies for User1
user = 'User1'
similar_users = similarity_df[user].sort_values(ascending=False)[1:]

# Weighted recommendation
recommendation_scores = df.T[similar_users.index].dot(similar_users) / similar_users.sum()
recommendations = recommendation_scores.sort_values(ascending=False)

print("Recommended Movies for User1:")
print(recommendations)
```

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Sample dataset (user-item ratings)
data = {
    'The Matrix': [4, 2, 5, 5, 5],
    'The Terminator': [4, 2, 4, 3, 4],
    'Pride and Prejudice': [3, 4, 3, 2, 2],
    'Malcom X': [5, 5, 5, 3, 2],
    'Enemy Mine': [5, 0, 5, 4, 0],
}

df = pd.DataFrame(data, index=['Father', 'Mother', 'Son #1', 'Son #2', 'Son #3'])

# Compute similarity between users
similarity = cosine_similarity(df)
similarity_df = pd.DataFrame(similarity, index=df.index, columns=df.index)

# Recommend movies for each user in the family
for user in df.index:
    print(f"\nRecommended Movies for {user}:")
    similar_users = similarity_df[user].sort_values(ascending=False)[1:]

    # Weighted recommendation
    recommendation_scores = df.T[similar_users.index].dot(similar_users) / similar_users.sum()
    recommendations = recommendation_scores.sort_values(ascending=False)

    print(recommendations)
```

An analysis of the code suggested that the *Recommended Movies* results are simply averaged numbers from the sample data located towards the beginning of the code, and, this curiosity prompted the question "Is the program averaging the numbers from the sample data set?" for Gemini AI to answer. The Gemini provided answer is "Yes, the program is essentially calculating a weighted average of the ratings from users similar to the target user. Let me explain the line: recommendation_scores = df.T[similar_users.index].dot(similar_users) / similar_users.sum()" (Gemini AI, 2026).

df.T[similar_users.index]: This part takes the movie ratings from the df DataFrame, transposes it, and then filters to include only the ratings from the users identified as 'similar users'.
.dot(similar_users): This performs a dot product. It multiplies each movie's rating from a similar user by that user's similarity score to the target user. This gives more weight to ratings from users who are more similar.
/ similar_users.sum(): Finally, this divides the sum of these weighted ratings by the total sum of the similarity scores. This normalization step results in a weighted average, where movies rated higher by more similar users get a higher overall recommendation score.
